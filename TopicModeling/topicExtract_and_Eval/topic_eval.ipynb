```python

client = OpenAI(api_key=APIKEY)
```

```python
def ask_chatgpt(prompt):
    response = client.chat.completions.create(
        model="gpt-5.4",
        response_format={"type": "json_object"},
        messages=[
            {
                "role": "system",
                "content": "You are an expert assistant for topic evaluetion. You must return valid JSON."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.0
    )

    return response.choices[0].message.content
```

```python
def build_prompt_granular(raw, report):
    prompt = f"""
Given the report  {report }, check the text analyzing and reporting the coverage across these topics, 
counting how many topics have been covered over the total number: {raw}
"""
    return prompt.strip()
```

```python
prompt_base = build_prompt_granular(raw,base)
output_base = ask_chatgpt(prompt_base)
```

```python
prompt_traj = build_prompt_granular(raw,traj)
output_traj = ask_chatgpt(prompt_traj)
```
