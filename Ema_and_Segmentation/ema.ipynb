```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
```

```python
df = pd.read_csv(r"C:\Users\Franco\Desktop\Pubblicazione\16\Kaggle_classification\output_Classification.csv", encoding="utf-8")
```

```python
df_clean=df[["Subject ID","Date","Text","Predicted_Label","Prob_Severe_Depressed","Prob_Moderate_Depressed","Prob_Not_Depressed"]]
```

```python
df_clean["prob"] = 2*df_clean["Prob_Severe_Depressed"]+df_clean["Prob_Moderate_Depressed"]

```

```python


subject_id = 2714      
df_u = df_clean[df_clean["Subject ID"] == subject_id]
```

```python


df_u["Date"] = pd.to_datetime(df_u["Date"]).dt.date
daily = df_u.groupby("Date")["prob"].mean().sort_index()

daily.index = pd.to_datetime(daily.index)
```

```python
h = 15
alpha = 0.5 ** (1 / h)

final_scores = []
prev = 0.0 

for base in daily:
    final = alpha * prev + (1 - alpha) * base
    final_scores.append(final)
    prev = final

daily_decayed = pd.Series(final_scores, index=daily.index)
serie = pd.Series(daily_decayed, index=daily.index)
```

```python
serie.to_csv(rf"C:\Users\Franco\Desktop\Pubblicazione\16\data\{subject_id}_score.csv")
```
